```python
import asyncio

async def task():
    num = 5;
    num2 = 10;
    total = int(num) + int(num2);
    print("Total :", total);
    await asyncio.sleep(1)
    print("update total is 20");
    
async def main():
    print("first addition")
    await task()

await main()
```

```python
async def task2(name):
    print(f"{name} started");
    await asyncio.sleep(2)
    print(f"{name} finished");

async def task3():
    await asyncio.gather(task2("Task 2"),task2("Task again 3"))

await task3()
```

```python
async def background():
    print("enter the data");
    await asyncio.sleep(5)
    print("sleep timing 1");

async def create():
    print("create data");
    asyncio.create_task(background())
    # await asyncio.create_task(background())
    await asyncio.sleep(2)
    print("finished")

await create()    
```
